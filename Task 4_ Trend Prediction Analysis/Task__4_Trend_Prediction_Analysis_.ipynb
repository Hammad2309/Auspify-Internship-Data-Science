# **Step 1: Prepare release-year data.**

```python
# importing libraries I need for this task
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# loading the cleaned dataset
df = pd.read_csv('/content/cleaned_netflix.csv')

# counting how many titles were released each year
yearly_counts = df['release_year'].value_counts().sort_index()

# creating a new dataframe to store year and count
yearly_df = pd.DataFrame({
    'year': yearly_counts.index,
    'count': yearly_counts.values
})

# checking if data is prepared properly
print("Yearly data is ready!")
print(yearly_df.head(10))
```

# **Step 2: Analyze yearly content trends.**

```python
# finding some basic insights from the data
print("Year with highest number of releases:", yearly_df[yearly_df['count'] == yearly_df['count'].max()]['year'].values[0])
print("Year with lowest number of releases:", yearly_df[yearly_df['count'] == yearly_df['count'].min()]['year'].values[0])
print("Average releases per year:", round(yearly_df['count'].mean(), 2))

# plotting a line chart to see the trend
plt.figure(figsize=(12,6))
plt.plot(yearly_df['year'], yearly_df['count'], marker='o', color='blue')
plt.title('Netflix Content Released Over the Years')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.grid(True)
plt.show()
```

# **Step 3: Build forecasting models.**

```python
# calculating 3-year moving average to see short-term trend
yearly_df['moving_avg_3'] = yearly_df['count'].rolling(window=3).mean()

# calculating 5-year moving average to see long-term trend
yearly_df['moving_avg_5'] = yearly_df['count'].rolling(window=5).mean()

# checking the last few rows to see if values are generated
print("Forecasting models created successfully!")
print(yearly_df.tail(10))
```

# **Step 4: Visualize future predictions.**

```python
# plotting actual vs predicted trends
plt.figure(figsize=(12,6))
plt.plot(yearly_df['year'], yearly_df['count'], label='Actual', marker='o', color='blue')
plt.plot(yearly_df['year'], yearly_df['moving_avg_3'], label='3-Year Moving Avg', linestyle='--', color='red')
plt.plot(yearly_df['year'], yearly_df['moving_avg_5'], label='5-Year Moving Avg', linestyle='--', color='green')
plt.title('Netflix Content Trends with Predictions')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.legend()
plt.grid(True)
plt.show()
```

# **Step 5: Interpret results.**

Trend Prediction Analysis(Results Summary)

1. Netflix has shown a clear upward trend in content addition over the years.
2. The highest number of titles was added in the year 2021.
3. The lowest number of titles was added in the early years (before 2010).
4. The 3-year moving average shows that the growth rate is still positive.
5. Based on historical data, it is expected that Netflix will continue adding more content in the coming years.
