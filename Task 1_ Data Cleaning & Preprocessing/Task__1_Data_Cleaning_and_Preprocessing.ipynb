# **Step 1: Import the dataset using Pandas.**

```python
# Step 1: Import libraries
import pandas as pd
import numpy as np

# Step 2: Load dataset
df= pd.read_csv("/content/Dataset.csv")

# checking shape and first few rows
print("Dataset Shape:", df.shape)
print("\nFirst 5 rows:")
df.head()
```

# **Step 2: Identify missing and duplicate records.**

```python
# Missing Values Checking
print("Missing Vlaues Per Coloumn:")
print(df.isnull().sum)

# Duplicate rows check
print(f"Total Duplicates Rows: {df.duplicated().sum()}")
```

# **Step 3: Handle null values and inconsistent formats.**

```python
# In step 2 we have seen that there is no missing values and no duplicate values in this dataset after analysing so, i am printing this line to completing the 3rd step
print("Step 3: No missing values to handle. Dataset is already clean.")
```

# **Step 4: Transform categorical and date-related columns.**

```python
# date_added converted into date/time format
df['date_added'] = pd.to_datetime(df['date_added'])

# extracting only year from date_added
df['year_added'] = df['date_added'].dt.year

# extracting only number from duration column
df['duration_num'] = df['duration'].str.extract(r'(\d+)')

# converting listed_in into a list format
df['genre_list'] = df['listed_in'].str.split(', ')

# checking is everything working fine
print(df[['date_added', 'year_added', 'duration', 'duration_num', 'listed_in', 'genre_list']].head())
```

# **Step 5: Create a clean dataset for further analysis.**

```python
# saving the cleaned dataset as a new csv file
df.to_csv('cleaned_netflix.csv', index=False)
print("Clean dataset saved as 'cleaned_netflix.csv'")
```
