# **# Step 1: Analyze dataset structure and statistics.**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
# loading the cleaned dataset
df = pd.read_csv('/content/cleaned_netflix.csv')

# checking dataset structure
print("Dataset Shape:", df.shape)
print("\nDataset Information:")
df.info()

# checking basic statistics
print("\nBasic Statistics for Numerical Columns:")
df.describe()
```

# **Step 2: Study content distribution by type.**

```python
# checking distribution of content types
type_counts = df['type'].value_counts()
print("Content Type Distribution:")
print(type_counts)

# calculating percentage
type_percentage = df['type'].value_counts(normalize=True) * 100
print("\nPercentage Distribution:")
print(type_percentage)
```

# **Step 3: Identify top countries and categories.**

```python
# top 10 countries with most content
top_countries = df['country'].value_counts().head(10)
print("Top 10 Countries by Content:")
print(top_countries)

# top 10 categories
top_genres = df['listed_in'].value_counts().head(10)
print("\nTop 10 Categories:")
print(top_genres)
```

# **Step 4: Create visualizations for key metrics.**

```python
# Pie chart for content type distribution
type_counts = df['type'].value_counts()
plt.figure(figsize=(8,6))
plt.pie(type_counts, labels=type_counts.index, autopct='%1.1f%%', startangle=90, colors=['skyblue', 'lightcoral'])
plt.title('Content Distribution: Movies vs TV Shows')
plt.show()
```

```python
# Bar chart for top 10 countries
top_countries = df['country'].value_counts().head(10)
plt.figure(figsize=(10,6))
sns.barplot(x=top_countries.values, y=top_countries.index, palette='viridis')
plt.title('Top 10 Countries by Content')
plt.xlabel('Number of Titles')
plt.show()
```

```python
# Bar chart for top 10 categories
top_genres = df['listed_in'].value_counts().head(10)
plt.figure(figsize=(10,6))
sns.barplot(x=top_genres.values, y=top_genres.index, palette='magma')
plt.title('Top 10 Categories/Genres')
plt.xlabel('Number of Titles')
plt.show()
```

```python
# Histogram for release year
plt.figure(figsize=(10,6))
sns.histplot(df['release_year'], bins=30, kde=True, color='green')
plt.title('Distribution of Release Years')
plt.xlabel('Release Year')
plt.ylabel('Frequency')
plt.show()
```

```python
# Histogram for duration (minutes)
plt.figure(figsize=(10,6))
sns.histplot(df['duration_num'], bins=30, kde=True, color='purple')
plt.title('Distribution of Duration (Minutes)')
plt.xlabel('Duration (Minutes)')
plt.ylabel('Frequency')
plt.show()
```

# **Step 5: Summarize findings.**

1. Netflix has more Movies (69.7%) than TV Shows (30.3%).
2. United States produces the most content (3248 titles), followed by India (1857).
3. Most content was added between 2018-2021, with 2021 having the highest number.
4. Drama, International Movies, and Documentaries are the most common categories.
5. The average movie duration is ~70 minutes, with most titles between 85-110 minutes.
